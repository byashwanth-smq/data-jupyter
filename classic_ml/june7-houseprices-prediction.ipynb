## Null check 

Either replace the Null/Na values to

- Mean or
- Drop or
- Most frequent element

```python
df.isnull().sum()
```

```python
df.isna().sum()
```

```python
df.columns
```

### Mean - to replace mean to few null columns

```python
df["LotFrontage"] = df["LotFrontage"].fillna(df["LotFrontage"].mean())
```

```python
df['GarageYrBlt'] = df['GarageYrBlt'].fillna(df['GarageYrBlt'].mean())
```

### Drop irrelavent columns
- if all row data of particular column is null

```python
df = df.drop(columns=['Alley', 'PoolQC', 'Fence', 'MiscFeature'])
```

### Hight frequency replacement (null check)

```python
categories = ['MasVnrType',"BsmtExposure", 'BsmtQual', 'BsmtCond', "BsmtExposure", "BsmtFinType2", "FireplaceQu", "GarageType", "GarageQual", "GarageCond", 'GarageFinish', 'BsmtFinType1']

for category in categories:
    [most_frequent_category] = df[category].mode()
    df[category] = df[category].fillna(most_frequent_category)
```

```python
df.isnull().sum()
```

### Drop very small nulls inplace

```python
df.dropna(inplace=True)
```
