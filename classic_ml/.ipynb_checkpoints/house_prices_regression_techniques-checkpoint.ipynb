```python
df["MasVnrType"] = df["MasVnrType"].fillna(df["MasVnrType"].mode()[0])
```

```python
df["MasVnrArea"] = df["MasVnrArea"].fillna(df["MasVnrArea"].mode()[0])
```

```python
df["LotFrontage"] = df["LotFrontage"].fillna(df["LotFrontage"].mode()[0])
```

```python
df["FireplaceQu"] = df["FireplaceQu"].fillna(df["FireplaceQu"].mode()[0])
```

```python
df.isnull().sum()
```

```python
cols = [
    "Utilities", "BsmtQual", "BsmtCond", "BsmtExposure",
    "BsmtFinType1", "BsmtFinSF1", "BsmtFinType2",
    "FireplaceQu", "GarageType", "GarageYrBlt",
    "GarageFinish", "GarageCars", "GarageArea",
    "GarageQual", "GarageCond"
]

for col in cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df.shape
```

```python
categories = ["MSZoning", "Street","LotShape", "LandContour", "LotConfig", "LandSlope", "Neighborhood", "Condition1", "Condition2", "BldgType", 
              "HouseStyle", "Exterior1st", "MasVnrType", "ExterQual", "ExterCond", "Foundation", "BsmtQual", "BsmtCond", "BsmtExposure", 
              "BsmtFinType1", "BsmtFinType2", "Heating", "HeatingQC", "CentralAir", "Electrical", "KitchenQual", "Functional", "FireplaceQu",
              "GarageType", "GarageFinish", "GarageQual", "GarageCond", "PavedDrive",  "SaleType", "SaleCondition", "RoofMatl", 
              "RoofStyle", "Exterior2nd", 
             ]
```

```python
for category in categories:
    splitted_category = pd.get_dummies(df[category], prefix=category)
    df = pd.concat([df, splitted_category], axis=1)
df = df.drop(columns=categories)
```

```python
df.head()
```

```python
df = df.drop(columns=["Id"])
```

```python
df.to_csv("updated_house_prices.py")
```

```python
splitted_rows = pd.get_dummies(df.Utilities)
df = pd.concat([df,splitted_rows], axis=1)
df = df.drop(columns=["Utilities"])
```
