```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("/Users/Yashwanth.B/jupyter/datasets/restaurant_tips.csv")
df.head()
```

```python
#Outliers removal

df.isnull().sum()
min_threshold , max_threshold = df['tip'].quantile([0.01, 0.95])
df2 = df[(df.tip> min_threshold) & (df.tip < max_threshold)]
print('Before qantile:',len(df))
print('After quantile:',len(df2))
```

```python
# Hot encoding -> breaking categories into number coulums based on type 

categories = ["sex", "smoker", "day", "time"]
for category in categories:
    divided_col = pd.get_dummies(df[category], prefix=category)
    df3 = pd.concat([df, divided_col], axis=1)
df3 = df3.drop(columns=categories)
df3.head()
```

```python
x = df3.drop(columns=['tip'])
y = df3['tip']
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x,y)
```

```python
df3 = df3.drop(columns=['tip'])
first_row = df3.iloc[[0]]
```

```python
output = model.predict(first_row)
output
```

```python
model.score(x, y)
```
