```python
import pandas as pd
import xgboost
```

```python
df = pd.read_csv("/Users/Yashwanth.B/jupyter/datasets/restaurant_tips.csv")
```

```python
df.head()
```

```python
df.isnull().sum() #checking for null values
```

```python
categories = ["sex", "smoker", "day", "time"]
```

```python
for category in categories:
    hotencoding_values = pd.get_dummies(df[category], prefix=category)
    df = pd.concat([df, hotencoding_values], axis=1)
df=df.drop(columns=categories)
print(df.head())
```

```python
x = df.drop(columns=["tip"])
y = df["tip"]
classifier = xgboost.XGBRegressor()
classifier.fit(x,y)
```

```python
df1 = df.drop(columns=["tip"])
first_row = df1.iloc[[0]]
```

```python
print(first_row)
```

```python
classifier.predict(first_row)
```
