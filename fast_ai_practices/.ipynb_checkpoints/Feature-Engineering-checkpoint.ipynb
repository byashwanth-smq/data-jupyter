### Newyork NYC Kaggle data analysis

```python
import pandas as pd
```

```python
df = pd.read_csv("folders/AB_NYC_2019.csv")
df.head()
```

```python
len(df)
```

```python
max_threshold = df['calculated_host_listings_count'].quantile(0.95)
```

```python
output = df[df['calculated_host_listings_count']>max_threshold]
len(output)
```

```python
min_threshold = df['calculated_host_listings_count'].quantile(0.05)
```

```python
filtered = df[(df['calculated_host_listings_count'] > min_threshold) & (df['calculated_host_listings_count'] < max_threshold)]
print(len(filtered))
```
